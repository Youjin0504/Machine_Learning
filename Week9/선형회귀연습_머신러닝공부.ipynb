```python
import numpy as np
from sklearn.linear_model import LinearRegression
```

```python
x = [1, 2, 3, 4, 5, 6, 7] #일주일에 머신러닝 공부라는 시간 (1차원이므로 x로 명명)
y = [47, 56, 66, 77, 85, 92, 98] #공부하는 시간에 따른 성적
```

```python
#모델은 선형 회귀인 LinearRegression
model = LinearRegression()
```

```python
#[참고] x 데이터에 .reshape(-1, 1)을 하는 이유는
#LinearRegression은 다중 선형 회귀 분석에도 사용되므로
#x의 변수가 하나가 아니라 여러개일 때 분석을 실시하기 위한 경우를 위해
#x는 2차원 array 형태여야 하기 떄문

x = np.array(x).reshape(-1,1)
```

```python
#모델 적합
model.fit(x, y)
```

```python
#기울기: model.coef_
print(model.coef_)
```

```python
#절편: model.intercept_
print(model.intercept_)
```

```python
#예측
n = input("머신러닝 일주일에 몇시간 공부할 건데? ")
print(model.predict([[float(n)]]))
```
