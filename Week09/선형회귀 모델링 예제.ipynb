```python
# Linear Regression 실습
# 아버지의 키와 아들의 키에대한 회귀모형
# x변수가 하나인 simple linear Regression
```

```python
import pandas as pd
import numpy as np
```

```python
# 데이터 로딩
data = pd.read_csv("./data/heights.csv")
data
```

```python
data.shape
```

```python
# 결측치 확인
data.isnull().sum()
```

```python
import matplotlib.pyplot as plt

X = data["father"]
y = data["son"]

plt.plot(X, y, '.')
plt.xlabel('Father')
plt.ylabel('Son')
```

```python
# 훈련데이터와 테스트데이터의 분할

from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state = 0)
```

```python
X_train.values
```

```python
# 훈련데이터(파란색)와 테스트데이터(붉은색) 분포 파악하기
plt.plot(X_train, y_train, 'b.', label = "train data")
plt.plot(X_test, y_test, 'r.', label = "test data")
plt.legend()
plt.xlabel('X train')
plt.ylabel('y train')
```

```python
# 모델 적합

from sklearn.linear_model import LinearRegression
lr = LinearRegression()

lr.fit(X_train.values.reshape(-1,1), y_train) # reshape(-1,1)으로 컬럼 갯수가 1개인 행렬로 조정
```

```python
# 회귀계수(기울기와 절편)
[lr.coef_, lr.intercept_]
```

```python
# 회귀모형을 통한 예측

# 예측함수 만들기 : y = b(절편) + a(기울기)x
f = lambda x: lr.intercept_ + (lr.coef_ * x)
```

```python
# f함수를 이용
f(180) #아버지 키가 180이라면?
```

```python
# 시각화
# train 데이터와 x값 범위(최소값, 최대값)
x = np.array([min(X_train), max(X_train)])
x
```

```python
plt.plot(X_train, y_train, 'b.') # train 데이터의 분포
plt.xlabel('X train')
plt.ylabel('Y train')
plt.plot(x, f(x), 'r') # 회귀식 그래프
```

```python
# 성능평가
```

```python
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score
```

```python
y_pred = lr.predict(X_test.values.reshape(-1,1))
y_pred
```

```python
mae = mean_absolute_error(y_test, y_pred)
mse = mean_squared_error(y_test, y_pred)
rmse = mse ** 0.5
```

```python
# 결정계수
r2 = r2_score(y_test, y_pred)
```

```python
print("MAE : %.2f" %mae)
print("MSE : %.2f" %mse)
print("RMSE : %.2f" %rmse)
print("R2 : %.2f" %r2)
```
